# 2. Weekly graph question

Below are the graphs from last week's weekly graph question.  Suppose you wanted to cause the viewer to examine the last three months of data first.  How could you use preattentive attributes to do that?  Suggest several possibilities.

```python
import numpy as np
import seaborn as sns
import pandas as pd
```

```python
numdates = 12
np.random.seed(0)
time_series = 100 * (np.random.normal(size = numdates) / 5 + np.arange(numdates) / 16)
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Nitrate level (% of eutrophic value)")
```

```python
np.random.seed(1)
time_series = 100 * (1 - (np.random.normal(size = numdates) / 8 + np.arange(numdates) / 16))
datearray = pd.date_range(start='2023/01/01', end='2024/01/01', freq='M')
ax = sns.lineplot(x = datearray, y = time_series)
ax.set_xlabel("Year and month")
ax.set_ylabel("Phosphate level (% of eutrophic value)")
```

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 99-119 of the Storytelling With Data book as best you can. (The first half of chapter four). You do not have to get the exact data values right, just the overall look and feel.

```python
# STORYTELLING WITH DATA GRAPH:

# Horizontal Bar Chart
import matplotlib.pyplot as plt
import seaborn as sns

categories =['Engine power is less than expected', 'Tires make excessive noise while driving', 'Engine makes abnormal/excessive noise', 'Seat material concerns', 'Excessive wind noise', 'Hesitation or delay when shifting', 'Bluetooth system has poor sound quality', 'Steering system/wheel has too much play', 'Bluetooth system is difficult to use', 'Front seat audio/entertainment/navigation controls']
values = [12.9, 12.3, 11.6, 11.6, 11.0, 10.3, 10.0, 8.8, 8.6, 8.2]

color_map = {
  'Engine power is less than expected': "pink",
  'Tires make excessive noise while driving': "red",
  'Engine makes abnormal/excessive noise': "red",
  'Seat material concerns': "pink",
  'Excessive wind noise': "red",
  'Hesitation or delay when shifting': "pink",
  'Bluetooth system has poor sound quality': "pink",
  'Steering system/wheel has too much play': "lightgrey",
  'Bluetooth system is difficult to use': "lightgrey",
  'Front seat audio/entertainment/navigation controls': "lightgrey"
}
colors = [color_map[c] for c in categories]

ax = sns.barplot(x=values, y=categories, color='lightgrey')  # initial plot

for spine in ax.spines.values():        # completely removes graph border 
    spine.set_visible(False)
    
for i, p in enumerate(ax.patches):      # set colors explicitly
    p.set_facecolor(colors[i])

# convert x-axis ticks to percentages
ax.set_xticklabels([f'{int(x)}%' for x in ax.get_xticks()])

# add data labels on top of each bar
for i, v in enumerate(values):
    ax.text(v + 0.5, i + 0.25, f'{v}%', color='grey', va='center')

# remove x-axis ticks
ax.set_xticks([])

plt.title("Top 10 design concerns")
plt.show()
```
