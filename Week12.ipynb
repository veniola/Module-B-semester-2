# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

### Graph 1 Advantages:
- Two Separate plots corresponding to each week (first week vs second week) 
- The same color is used consistently 
- Both x and y axes start at zero
### Graph 1 Disadvantages:
- Too much clutter--there's not enough spacing between the two graphs
- There is a shared y-axis between the two graphs although the first graph has its own y-axis, and each y-axis value is different
- Axis titles are placed in terrible spots (middle of graph and bottom right)
### Graph 2 Advantages:
- A shared plot with two lines corresponding to each week (first week vs second week) 
- Great use of color to distinguish between each week (first week vs second week)
- Axis titles are well placed 
### Graph 2 Disadvantages:
- There is no legend to distinguish the colors (corresponding week) from each other
- The graph is misisng a descriptive title
- No other disadvantages

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
# STORYTELLING WITH DATA GRAPH:

# Average Retail Product Price Per Year chart
import matplotlib.pyplot as plt
import seaborn as sns

metrics = ["Product A", "Product B", "Product C", "Product D", "Product E"]
yr19 = [390, 360, 0, 0, 0]
yr20 = [420, 400, 0, 0, 0]
yr21 = [430, 410, 100, 0, 0]
yr22 = [380, 370, 185, 170, 0]
yr23 = [300, 290, 200, 270, 0]
yr24 = [260, 250, 245, 235, 100]
yr25 = [255, 245, 185, 215, 215]

x = np.arange(len(metrics))
palette = sns.color_palette(["darkblue", "red", "green", "purple", "lightblue", "orange", "grey"])
# Build a DataFrame where each column is a year and each row is a product
df_years = pd.DataFrame({
    '2019': yr19,
    '2020': yr20,
    '2021': yr21,
    '2022': yr22,
    '2023': yr23,
    '2024': yr24,
    '2025': yr25
}, index=metrics)

# Plot grouped bars (side-by-side) so each product shows one bar per year
ax = df_years.plot(kind='bar', stacked=False, figsize=(20,15), color=palette, linewidth=0.6, width=0.8)

# Axes, grid, legend
ax.set_xticks(x)
ax.set_xticklabels(metrics, rotation=0, fontsize=14)
ax.set_yticklabels([f'${y}' for y in ax.get_yticks()], rotation=0, fontsize=14) # add $ to y-axis labels
ax.set_title('Average Retail Product Price Per Year', fontsize=16)
ax.legend(frameon=True, loc='center', bbox_to_anchor=(0.5, -0.05), ncol=7, fontsize=14)
ax.grid(axis='y', linestyle='--', alpha=0.6, zorder=0)



#plt.tight_layout()
plt.show()
```
